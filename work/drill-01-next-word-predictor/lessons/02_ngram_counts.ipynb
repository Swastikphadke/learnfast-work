```python
CORPUS = """I drink coffee every morning before work. The morning train to the city is often late. She reads a book on the train to work. We drink tea in the afternoon at the office. The weather is cold and grey this morning. He walks to the station in the rain. On the weekend we walk in the park. The park is quiet and green in the morning. I read a good book on the weekend. Coffee in the morning helps me wake up. The city is busy and loud during the week. She takes the late train home from work. We cook dinner together on the weekend. The rain falls softly on the quiet street. He drinks his coffee black and strong. The office is warm and bright in the winter. I walk to work when the weather is nice. The train was late again this morning. She writes in her journal every evening. We watch a film together on friday night. The dog runs happily in the green park. Morning coffee and a good book make me happy. The weekend is short but the week is long. He reads the news on his phone every morning. The garden is full of flowers in the spring. I like a quiet morning with hot coffee. The children play in the park after school. She cooks a warm dinner on cold evenings. We travel to the coast in the summer. The sea is calm and blue in the morning. Rainy mornings make me want more coffee. The old cafe on the corner makes good coffee. He waits for the train on the cold platform. I finish work and walk home in the evening. The city lights are bright at night. She plays the piano softly in the evening. We read together by the warm fire in winter. The bakery smells of fresh bread every morning. I buy fresh bread and coffee on the weekend. The bus is slower than the morning train. He enjoys a long walk on sunny weekends. The library is quiet and full of old books. She borrows a new book from the library. Cold weather makes the morning coffee taste better. We plan a short trip for the long weekend. The mountains are beautiful in the early morning. I wake up early and make strong coffee. The evening is calm after a busy week. He cooks pasta for dinner on friday. The rain stopped and the sun came out. We sit in the park and read our books. Morning trains are crowded during the week. She drinks green tea instead of coffee. The winter mornings are dark and cold. I love the smell of coffee in the morning. The weekend market sells fresh fruit and bread. He reads a chapter of his book before bed. The quiet street is empty late at night. We walk the dog in the park every evening. Good coffee and a good book start my day."""

with open("corpus.txt", "w") as f:
    f.write(CORPUS)

print("Saved corpus.txt —", len(CORPUS), "characters")
```

```python
import re
import math
import random

from collections import Counter, defaultdict


def to_sentences(text):
    return [
        s.strip()
        for s in re.split(r"[.!?]+", text)
        if s.strip()
    ]


def to_tokens(sentence):
    return re.findall(r"[a-z']+", sentence.lower())


sents = to_sentences(CORPUS)

tokens = [
    token
    for sentence in sents
    for token in to_tokens(sentence)
]

print("Sentences:", len(sents))
print("Tokens:", len(tokens))
print("Vocabulary:", len(set(tokens)))

print("\nFirst sentence:", sents[0])
print("Its tokens:", to_tokens(sents[0]))
```

```python
def build(sents, n):
    """
    context (n-1 words) -> Counter of following words
    """

    foll = defaultdict(Counter)

    for sentence in sents:

        toks = (
            ["<s>"] * (n - 1)
            + to_tokens(sentence)
            + ["</s>"]
        )

        for i in range(len(toks) - n + 1):

            context = tuple(toks[i:i + n - 1])
            next_word = toks[i + n - 1]

            foll[context][next_word] += 1

    return foll
```

```python
foll = build(sents, n=2)

print("Words that follow 'the':")
print(foll[("the",)].most_common(6))

print("\nWords that follow 'morning':")
print(foll[("morning",)].most_common(5))

print("\nWords that start a sentence:")
print(foll[("<s>",)].most_common(5))
```

```python
trigrams = build(sents, n=3)

print("Words following 'in the':")
print(trigrams[("in", "the")].most_common(5))

print("\nWords following 'on the':")
print(trigrams[("on", "the")].most_common(5))

print("\nWords following 'good book':")
print(trigrams[("good", "book")].most_common(5))
```

```python
print("BIGRAM:")
print(foll[("the",)].most_common(10))

print("\nTRIGRAM:")
print(trigrams[("in", "the")].most_common(10))
```

## Observations

1. A corpus is a collection of text used to train or analyse a language model.
2. Tokenization converts sentences into individual words.
3. Lowercasing ensures consistent word representation.
4. Special tokens <s> and </s> represent sentence boundaries.
5. Bigram models use one previous word as context.
6. Trigram models use two previous words as context.
7. Counter stores the frequency of following words.
8. defaultdict automatically creates a Counter for unseen contexts.
9. Increasing n makes the context more specific.
10. This implementation calculates raw counts, not probabilities.
