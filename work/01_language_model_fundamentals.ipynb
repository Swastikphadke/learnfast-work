# Lesson 1: What a Language Model Is

## Objective
Understand how language models predict the next word using
probability distributions and limited word context.

A language model estimates:

P(next word | previous words)

```python
next_word = {
    "the": {
        "morning": 0.30,
        "train": 0.20,
        "park": 0.15,
        "city": 0.10,
        "weekend": 0.08
    },
    "good": {
        "book": 0.50,
        "coffee": 0.50
    }
}

print(next_word)
```

```python
def suggest(context, table, k=3):
    ranked = sorted(
        table.get(context, {}).items(),
        key=lambda kv: -kv[1]
    )
    return ranked[:k]


print("After 'the':", suggest("the", next_word))
print("After 'good':", suggest("good", next_word))
```

```python
print("Top 1:", suggest("the", next_word, k=1))
print("Top 5:", suggest("the", next_word, k=5))
print("Unknown:", suggest("hello", next_word))
```

## Observations

1. A language model assigns probabilities to possible next words.
2. Autocomplete displays the most probable suggestions.
3. Text generation repeatedly predicts the next word.
4. A bigram model uses one previous word as context.
5. A trigram model uses two previous words as context.
6. The Markov assumption limits the context to make prediction manageable.

The probabilities in this experiment were manually assigned.
